```python
import pandas as pd
```

```python
df = pd.read_csv("car_ownership.csv")
df.head(5)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Step 1: Features and target
X = df[['monthly_salary']]  # double brackets for DataFrame
y = df['owns_car']

# Step 2: Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)

# Step 3: Model training
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
X_test.head()
```

```python
model.predict(X_test)
```

```python
y_test.tolist()
```

```python
model.predict_proba(X_test)
```

```python
model.score(X_test, y_test)
```

```python
model.coef_, model.intercept_
```

```python
import math

def sigmoid(Z):
    y = 1/ (1 + math.exp(-Z))
    return y

def prediction_func(X):
    Z = model.coef_ * X + model.intercept_
    y = sigmoid(Z)
    return y

prediction_func(62000)
```

```python
import math

def prediction_func(X, model):
    Z = model.coef_ * X + model.intercept_
    y = 1/ (1 + math.exp(-Z))
    return y

prediction_func(62000, model)
```
